```python
import numpy as np

# Use your first 20 categories
first_20_cats = categories[:20]

# Store average strokes
avg_strokes = {}

for cat in first_20_cats:
    print(f"Processing: {cat}")
    qdg = QuickDrawDataGroup(cat, recognized=True, max_drawings=1000)

    strokes_per_drawing = []
    for i, drawing in enumerate(qdg.drawings):
        if i >= 100:  # Limit for speed
            break
        strokes_per_drawing.append(len(drawing.strokes))

    if strokes_per_drawing:
        avg_strokes[cat] = np.mean(strokes_per_drawing)
    else:
        avg_strokes[cat] = 0

# Display the results
for cat, avg in avg_strokes.items():
    print(f"{cat}: Avg Strokes = {avg:.2f}")
```

```python

aspect_ratios = []

for cat in categories[:20]:  # First 20 categories
    pattern = os.path.join(cat, "*.png")
    files = glob.glob(pattern)

    for f in files[:100]:  # Sample 100 per category
        img = Image.open(f).convert("L")
        w, h = img.size
        if h > 0:  # Avoid division by zero
            aspect_ratios.append(w / h)

# Plot aspect ratio histogram
plt.figure(figsize=(8, 4))
plt.hist(aspect_ratios, bins=20, color='slateblue', edgecolor='black')
plt.title("Aspect Ratio Distribution (Width / Height)")
plt.xlabel("Aspect Ratio")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
incomplete_counts = {}

threshold = 100  # Number of pixels below which we call it "incomplete"

for cat in categories:  # limit to first 20 for now
    pattern = os.path.join(cat, "*.png")
    files = glob.glob(pattern)

    incomplete = 0

    for f in files:
        img = Image.open(f).convert("L")
        arr = np.array(img)

        # Count how many pixels are 'dark' (less than 200, meaning drawn)
        dark_pixels = np.sum(arr < 200)

        if dark_pixels < threshold:
            incomplete += 1

    incomplete_counts[cat] = incomplete

# Display results
for cat, count in incomplete_counts.items():
    print(f"{cat}: {count} likely incomplete drawings")
```

```python
# Sort or just take the first 20 if unsure
limited_counts = dict(list(incomplete_counts.items())[:20])

# Prepare data
cats = list(limited_counts.keys())
counts = list(limited_counts.values())

# Plot
plt.figure(figsize=(10, 5))
plt.bar(cats, counts, color='crimson')
plt.xticks(rotation=90)
plt.ylabel("Number of Likely Incomplete Drawings")
plt.title("Incomplete Drawing Counts (First 20 Categories)")
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import pandas as pd

whiteness_data = []

for cat in categories[:20]:
    pattern = os.path.join(cat, "*.png")
    files = glob.glob(pattern)

    for f in files[:100]:  # limit for speed
        img = Image.open(f).convert("L").resize((28, 28))
        arr = np.array(img)

        # Calculate percentage of white-ish pixels (very close to 255)
        white_pixels = np.sum(arr > 250)
        total_pixels = arr.size
        whiteness_pct = (white_pixels / total_pixels) * 100

        whiteness_data.append({
            "Category": cat,
            "Whiteness (%)": whiteness_pct
        })

# Convert to DataFrame
df_white = pd.DataFrame(whiteness_data)

# Create boxplot
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_white, x="Category", y="Whiteness (%)", color="lightgray")
plt.xticks(rotation=90)
plt.title("Distribution of Whiteness per Drawing (First 20 Categories)")
plt.tight_layout()
plt.show()
```
