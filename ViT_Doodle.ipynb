```python
import os
from PIL import Image

# Paths
SOURCE_DIR = "/content/"  # current directory where images are stored in category folders
DEST_DIR = "/content/"
RESIZE_SHAPE = (224, 224)

# Get all category folders
categories = [folder for folder in os.listdir(SOURCE_DIR) if os.path.isdir(os.path.join(SOURCE_DIR, folder))]

for category in categories:
    src_category_dir = os.path.join(SOURCE_DIR, category)
    dest_category_dir = os.path.join(DEST_DIR, category)

    os.makedirs(dest_category_dir, exist_ok=True)

    print(f"Processing category: {category}")

    for filename in os.listdir(src_category_dir):
        if filename.endswith(".png"):
            src_path = os.path.join(src_category_dir, filename)
            dest_path = os.path.join(dest_category_dir, filename)

            try:
                with Image.open(src_path) as img:
                    img = img.convert("RGB")
                    img = img.resize(RESIZE_SHAPE, Image.Resampling.LANCZOS)
                    img.save(dest_path)
            except Exception as e:
                print(f"Error processing {src_path}: {e}")

    print(f"✅ Saved resized images to {dest_category_dir}\n")
```

```python
import os
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from transformers import ViTForImageClassification
from torch.optim import AdamW
import random
import torch
from tqdm import tqdm
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
```

```python
# Step 1: Set device (GPU or CPU)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Step 2: Define the transformations
transform = transforms.Compose([
    transforms.Resize((224, 224)),  # Resize images to 224x224 for ViT input
    transforms.ToTensor(),  # Convert images to tensor
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),  # Normalize using ImageNet mean and std
])

# Step 3: Define the valid image extensions
valid_extensions = {'.jpg', '.jpeg', '.png', '.ppm', '.bmp', '.pgm', '.tif', '.tiff', '.webp'}

# Step 4: Filter dataset directory to include only valid image files
dataset_dir = '/content/'  # Replace with your dataset directory
```

```python
# Create an empty list to hold valid image paths and corresponding labels
valid_images = []
valid_labels = []

# Loop through the directory and find valid image files
for class_name in os.listdir(dataset_dir):
    class_path = os.path.join(dataset_dir, class_name)

    # Ensure we are only looking at directories (classes)
    if os.path.isdir(class_path):
        for file_name in os.listdir(class_path):
            file_path = os.path.join(class_path, file_name)
            if os.path.splitext(file_path)[1].lower() in valid_extensions:  # Check if file is a valid image
                valid_images.append(file_path)
                valid_labels.append(class_name)  # The folder name is the label

# Step 5: Randomly select 20% of the images for training
num_images_to_select = int(0.2 * len(valid_images))  # 20% of the entire dataset
selected_indices = random.sample(range(len(valid_images)), num_images_to_select)

# Get the selected images and labels using the random indices
selected_images = [valid_images[i] for i in selected_indices]
selected_labels = [valid_labels[i] for i in selected_indices]
```

```python
# Step 6: Create a custom dataset using the filtered images and labels
class CustomDataset(torch.utils.data.Dataset):
    def __init__(self, image_paths, labels, transform=None, unique_classes=None):
        self.image_paths = image_paths
        self.labels = labels
        self.transform = transform
        self.unique_classes = unique_classes

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image_path = self.image_paths[idx]
        label = self.labels[idx]

        # Load the image
        image = datasets.folder.default_loader(image_path)

        if self.transform:
            image = self.transform(image)

        # Return image and label (mapping label to index in unique_classes)
        return image, self.unique_classes.index(label)  # Map class label to an index in the unique_classes list
```

```python
# Step 7: Create DataLoader
unique_classes = list(set(valid_labels))  # Get the unique class names from the entire dataset
train_dataset = CustomDataset(selected_images, selected_labels, transform=transform, unique_classes=unique_classes)
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)

# Step 8: Initialize the ViT model
model = ViTForImageClassification.from_pretrained("google/vit-base-patch16-224-in21k")

# Modify the classifier to match the number of classes in your dataset
num_classes = len(unique_classes)
model.classifier = torch.nn.Linear(model.config.hidden_size, num_classes)

# Move model to the correct device (GPU or CPU)
model.to(device)
```

```python
from sklearn.model_selection import train_test_split

# Split the selected images into training and validation sets
train_imgs, val_imgs, train_lbls, val_lbls = train_test_split(
    selected_images, selected_labels, test_size=0.2, stratify=selected_labels
)

# Create PyTorch datasets
train_dataset = CustomDataset(train_imgs, train_lbls, transform=transform, unique_classes=unique_classes)
val_dataset   = CustomDataset(val_imgs, val_lbls, transform=transform, unique_classes=unique_classes)

# Create DataLoaders
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader   = DataLoader(val_dataset, batch_size=32, shuffle=False)
```

```python
# Step 9: Define the optimizer
optimizer = AdamW(model.parameters(), lr=1e-5)

# Step 10: Define loss function
criterion = torch.nn.CrossEntropyLoss()

# Early stopping parameters
min_delta = 0.01
best_val_loss = float('inf')
patience = 5
patience_counter = 0

num_epochs = 15

for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0
    correct_preds = 0
    total_preds = 0

    for images, labels in tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs}", leave=False):
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs.logits, labels)
        running_loss += loss.item()
        loss.backward()
        optimizer.step()

        _, preds = torch.max(outputs.logits, 1)
        correct_preds += torch.sum(preds == labels).item()
        total_preds += labels.size(0)

    epoch_loss = running_loss / len(train_loader)
    epoch_acc = correct_preds / total_preds * 100

    # --- Validation ---
    model.eval()
    val_loss = 0.0
    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss = criterion(outputs.logits, labels)
            val_loss += loss.item()

    avg_val_loss = val_loss / len(val_loader)
    print(f"Epoch [{epoch+1}/{num_epochs}] - Train Loss: {epoch_loss:.4f}, Accuracy: {epoch_acc:.2f}% | Val Loss: {avg_val_loss:.4f}")

    # --- Early Stopping Check ---
    if (best_val_loss - avg_val_loss) > min_delta:
        best_val_loss = avg_val_loss
        patience_counter = 0
        torch.save(model.state_dict(), 'best_model.pt')
        print("✅ Validation loss improved. Model saved.")
    else:
        patience_counter += 1
        print(f"⏳ No significant improvement. Patience: {patience_counter}/{patience}")
        if patience_counter >= patience:
            print("🛑 Early stopping triggered.")
            break
```

```python
# Step 12: Save the model after training
torch.save(model.state_dict(), 'vit_model.pth')
print("Model saved as vit_model.pth")

# --------------------- 9.  EVALUATION ---------------------------------------
model.eval()
all_preds, all_true = [], []
with torch.no_grad():
    for imgs, lbls in val_loader:
        imgs, lbls = imgs.to(device), lbls.to(device)
        logits     = model(imgs).logits
        all_preds.extend(torch.argmax(logits, 1).cpu().numpy())
        all_true.extend(lbls.cpu().numpy())

print("\n" + classification_report(all_true, all_preds,
                                   target_names=unique_classes, digits=4))
```
